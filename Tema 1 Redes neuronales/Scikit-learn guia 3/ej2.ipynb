```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split, KFold
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
#cargo los datos
digits = load_digits()
X = digits.data
y = digits.target
```

```python
#creo los clasificadores
modelos = {
    "MLP": MLPClassifier(hidden_layer_sizes=(100,),max_iter=1000,random_state=42),
    "Naive Bayes": GaussianNB(),
    "LDA": LinearDiscriminantAnalysis(),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Árbol de decisión": DecisionTreeClassifier(random_state=42),
    "SVM": SVC()
}
```

```python
#hago el fold de 5 para comparar
kf = KFold(n_splits=5,shuffle=True,random_state=42)
```

```python
resultados = {
    nombre: []
    for nombre in modelos
}
print(resultados)
```

```python
#entrenamos y evaluamos
#aca lo q hacemos es iterar por los fold y hacerle el train y test
#a cada fold para cada clasificador una vez hacemos eso
#pasamos al otro fold y asi sucesivamente
for fold, (train_idx, test_idx) in enumerate(kf.split(X)):
    
    X_train = X[train_idx]
    y_train = y[train_idx]
    
    X_test = X[test_idx]
    y_test = y[test_idx]
    
    print(f"\nFold {fold + 1}")
    
    for nombre, modelo in modelos.items():
        
        modelo.fit(X_train, y_train)
        
        precision = modelo.score(X_test, y_test)
        
        resultados[nombre].append(precision)
        
        print(f"{nombre}: {precision:.4f}")
```

```python
#calculamos medias y varianzas
medias = {}
varianzas = {}

for nombre, accuracies_modelo in resultados.items():
    medias[nombre] = np.mean(accuracies_modelo)
    varianzas[nombre] = np.var(accuracies_modelo)

    print(f"{nombre}:")
    print(f"  Media: {medias[nombre]:.4f}")
    print(f"  Varianza: {varianzas[nombre]:.6f}")
```

```python
#conclusion
#se observa que SVM obtuvo la mayor tasa de acierto promedio, con un 98,78 %, 
# seguido muy de cerca por KNN, con un 98,61 %. 
# El MLP obtuvo un 96,94 %, 
# mientras que LDA alcanzó un 95,21 %. 
# Los menores resultados correspondieron a Árbol de decisión y Naive Bayes, con 85,70 % y 83,92 %, respectivamente. 
# SVM también presentó la menor varianza, indicando una gran estabilidad entre las distintas particiones. 
# Por lo tanto, para este conjunto de datos, SVM fue el clasificador que presentó el mejor desempeño general.
```
